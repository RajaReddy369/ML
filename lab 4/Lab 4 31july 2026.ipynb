```python
import numpy as np
from hmmlearn import hmm

# ==========================================
# Task 1: Forward Algorithm from Scratch
# ==========================================

# 1. Model Parameters
states = ["Walking", "Running", "Resting"]
observations = ["Low", "Medium", "High"]

# Initial State Probabilities (\pi)
pi = np.array([0.5, 0.3, 0.2])

# Transition Matrix (A)
A = np.array([
    [0.6, 0.3, 0.1],  # Walking
    [0.2, 0.5, 0.3],  # Running
    [0.3, 0.2, 0.5]   # Resting
])

# Emission Matrix (B)
B = np.array([
    [0.6, 0.3, 0.1],  # Walking
    [0.1, 0.3, 0.6],  # Running
    [0.7, 0.2, 0.1]   # Resting
])

obs_map = {"Low": 0, "Medium": 1, "High": 2}
seq_text = ["High", "Medium", "Low", "High"]
seq = [obs_map[o] for o in seq_text]

def forward_algorithm(pi, A, B, seq):
    N = len(pi)
    T = len(seq)
    alpha = np.zeros((T, N))
    
    # Step 1: Initialization
    alpha[0] = pi * B[:, seq[0]]
    
    # Step 2: Induction
    for t in range(1, T):
        for j in range(N):
            alpha[t, j] = np.sum(alpha[t-1] * A[:, j]) * B[j, seq[t]]
            
    # Step 3: Termination
    likelihood = np.sum(alpha[-1])
    return alpha, likelihood

alpha_table, likelihood_t1 = forward_algorithm(pi, A, B, seq)

print("=== TASK 1 RESULTS ===")
print("Forward Probability Table (Alpha):")
print("Time Step | Walking   | Running   | Resting")
print("-" * 43)
for t, row in enumerate(alpha_table):
    print(f"t = {t+1} ({seq_text[t]:<6}) | {row[0]:.6f}  | {row[1]:.6f}  | {row[2]:.6f}")

print(f"\nLikelihood of sequence O {seq_text}: {likelihood_t1:.8f}\n")


# ==========================================
# Task 2: HAR using hmmlearn
# ==========================================

print("=== TASK 2 RESULTS ===")

# Task 2.1 & 2.2: Construct CategoricalHMM
model = hmm.CategoricalHMM(n_components=3, init_params="")
model.startprob_ = pi
model.transmat_ = A
model.emissionprob_ = B

# Task 2.3 & 2.4: Observation Likelihoods
sequences = {
    "O": ["High", "Medium", "Low", "High"],
    "O1": ["Low", "Medium", "High", "Medium"],
    "O2": ["High", "High", "High", "High"],
    "O3": ["Low", "Low", "Low", "Low"],
    "O4": ["Medium", "Medium", "Medium", "Medium"]
}

print("--- Task 2.3 & 2.4: Observation Likelihoods ---")
for name, seq_labels in sequences.items():
    obs_indices = np.array([[obs_map[o]] for o in seq_labels])
    log_likelihood = model.score(obs_indices)
    likelihood = np.exp(log_likelihood)
    print(f"{name} {seq_labels}:")
    print(f"  Log-Likelihood : {log_likelihood:.6f}")
    print(f"  Likelihood     : {likelihood:.8f}")

# Task 2.6: Predict Most Probable Hidden State Sequence (Viterbi Algorithm)
obs_indices_O = np.array([[obs_map[o]] for o in seq_text])
log_prob, state_sequence = model.decode(obs_indices_O, algorithm="viterbi")
predicted_states = [states[s] for s in state_sequence]

print("\n--- Task 2.6: Most Probable Hidden State Sequence ---")
print(f"Observation Sequence : {seq_text}")
print(f"Predicted States     : {predicted_states}")


# Task 2.7 & 2.8: Train Model using Baum-Welch Algorithm
print("\n--- Task 2.7 & 2.8: Baum-Welch Training & Comparison ---")

# Construct a trainable model
trained_model = hmm.CategoricalHMM(n_components=3, n_iter=100, random_state=42)

# Train using Baum-Welch (EM) on sequence O
trained_model.fit(obs_indices_O)

print("\n[i] Learned Initial State Vector (pi):")
print(trained_model.startprob_.round(4))

print("\n[ii] Learned Transition Matrix (A):")
print(trained_model.transmat_.round(4))

print("\n[iii] Learned Emission Matrix (B):")
print(trained_model.emissionprob_.round(4))

print("\n[iv] Parameter Comparison & Observations:")
print("- The original model parameters were based on broad prior knowledge across all activities.")
print("- The trained model adjusts parameters to fit *only* the single sequence O = (High, Medium, Low, High).")
print("- Due to the extremely short dataset (4 observations), the trained parameters become overfitted.")
```
